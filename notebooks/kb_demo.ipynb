```python
# phase 1 Measure OCR Quality on a Sample

import os
import jsonlines
import easyocr
from jiwer import cer

# 1. Define paths to your grading kit data
LABELS_PATH = "../grading_kit/labels.jsonl"
PAGES_DIR = "../grading_kit/heldout_pages"

# Lists to store ground truth and OCR predictions
ground_truths = []
predictions = []

print("Starting OCR evaluation using EasyOCR (bn+en) on 9 held-out pages...\n")

# 2. Initialize EasyOCR (it will download the bn and en model weights the first time you run this)
# Setting gpu=False ensures it runs on CPU if you don't have a configured GPU
reader = easyocr.Reader(['bn', 'en'], gpu=False)

# 3. Open and read the labels.jsonl file
try:
    with jsonlines.open(LABELS_PATH) as f:
        for obj in f:
            file_name = obj.get("page_id")
            true_text = obj.get("text", "").strip()
            
            # Construct the full path to the image
            image_path = os.path.join(PAGES_DIR, file_name)
            
            # Skip if file doesn't exist
            if not os.path.exists(image_path):
                print(f"Warning: Image {file_name} not found in {PAGES_DIR}")
                continue
                
            # 4. Run EasyOCR
            # detail=0 returns just a list of extracted text strings (ignoring bounding box coordinates for now)
            result = reader.readtext(image_path, detail=0)
            
            # Join the strings with a space (as Abdullah noted, EasyOCR doesn't do newlines well)
            pred_text = " ".join(result).strip()
            
            # 5. Store the results
            ground_truths.append(true_text)
            predictions.append(pred_text)
            
            print(f"✅ Processed {file_name}")
            
except FileNotFoundError:
    print(f"❌ Error: Could not find {LABELS_PATH}")

# 6. Calculate overall Character Error Rate (CER)
if ground_truths and predictions:
    overall_cer = cer(ground_truths, predictions)
    
    print("\n" + "="*40)
    print("🎯 OCR EVALUATION RESULTS (EasyOCR)")
    print("="*40)
    print(f"Total Pages Evaluated: {len(ground_truths)}")
    print(f"Overall Character Error Rate (CER): {overall_cer:.4f}")
    
    # Calculate Accuracy as (1 - CER)
    accuracy = max(0, 1 - overall_cer)
    print(f"Estimated Character Accuracy: {accuracy:.2%}")
else:
    print("No data was evaluated. Check your labels.jsonl file.")
```

```python
import faiss
import json
import os
import torch
import jsonlines
from sentence_transformers import SentenceTransformer

# 1. DEFINE KAGGLE PATHS
INDEX_PATH = "/kaggle/input/datasets/prachurjadhar/index-physics/index-pilot/index.faiss"
METADATA_JSON_PATH = "/kaggle/input/datasets/prachurjadhar/index-physics/index-pilot/metadata.json"
CHUNKS_JSONL_PATH = "/kaggle/input/datasets/prachurjadhar/index-physics/index-pilot/chunks.jsonl"

print("Loading files...")
if not os.path.exists(INDEX_PATH):
    print(f"❌ Error: Could not find {INDEX_PATH}.")
else:
    # --- PHASE 2: INDEX STATISTICS ---
    print("Loading FAISS index...")
    index = faiss.read_index(INDEX_PATH)
    
    print("Loading Chunk Metadata...")
    metadata = []
    
    if os.path.exists(CHUNKS_JSONL_PATH):
        with jsonlines.open(CHUNKS_JSONL_PATH) as f:
            for obj in f:
                metadata.append(obj)
    elif os.path.exists(METADATA_JSON_PATH):
        with open(METADATA_JSON_PATH, "r", encoding="utf-8") as f:
            metadata = json.load(f)
            if isinstance(metadata, dict) and "chunks" in metadata:
                metadata = metadata["chunks"]
    
    total_chunks = index.ntotal
    embedding_dimension = index.d
    
    # Calculate corpus coverage
    unique_pages = set()
    total_words = 0
    for chunk in metadata:
        page_id = chunk.get("page_id") or chunk.get("page_no") or "unknown"
        text = chunk.get("text", "")
        words = chunk.get("n_words", len(text.split()))
        
        unique_pages.add(page_id)
        total_words += words

    print("\n" + "="*40)
    print("📊 PHASE 2: INDEX STATISTICS")
    print("="*40)
    print(f"Total Chunks Indexed: {total_chunks:,}")
    print(f"Embedding Dimension: {embedding_dimension}")
    print(f"Index Type: FAISS IndexFlatIP")
    print(f"Corpus Coverage: {len(unique_pages)} pages / {total_words:,} words")

    # --- PHASE 3: RETRIEVAL ---
    print("\n" + "="*40)
    print("🔍 PHASE 3: ONE RETRIEVAL EXAMPLE")
    print("="*40)

    # 1. Load the embedding model onto Kaggle's GPU
    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    print(f"Loading 'multilingual-e5-large' on: {device.upper()}...")
    embedder = SentenceTransformer('intfloat/multilingual-e5-large', device=device)

    # 2. Define a Physics query
    test_query = "কাজ ও শক্তির মধ্যে সম্পর্ক কী?" # "What is the relationship between work and energy?"
    print(f"Query: '{test_query}'")

    # 3. Embed the query
    query_embedding = embedder.encode([f"query: {test_query}"], normalize_embeddings=True)

    # 4. Search the FAISS index
    k = 1 # Top 1 result
    distances, indices = index.search(query_embedding, k)

    # 5. Display the result
    top_idx = indices[0][0]
    score = distances[0][0]
    
    retrieved_chunk = metadata[top_idx]
    retrieved_text = retrieved_chunk.get("text", "No text found")
    page_id = retrieved_chunk.get("page_id") or retrieved_chunk.get("page_no") or "Unknown page"

    print(f"\n✅ Top Match Score (Cosine Similarity): {score:.4f}")
    print(f"📄 Found on Page: {page_id}")
    print(f"📝 Retrieved Chunk:\n{retrieved_text}")
```
